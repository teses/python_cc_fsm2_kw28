# Eingabe einer Zahl mit Datentypwandlung (Validierung)

```python
while True:
    try:
        eingabe = int(input("Bitte geben Sie eine Zahl ein: "))
        break
    except:
        print("Bitte geben Sie eine gültige ganze Zahl ein. ")


print(eingabe)
```

# Eingabe einer Kommazahl mit Datentypwandlung (Validierung)

```python
while True:
    try:
        eingabe = float(input("Bitte geben Sie eine kommazahl an: "))
        break
    except Exception as e:
        print("Bitte geben Sie eine gültige kommazahl ein. ")

print(eingabe)
```

# Eingabe einer Zahl mit Datentypwandlung (Validierung) im Bereich  1 bis 6

```python
while True:
    try:
        eingabe = int(input("Bitte geben Sie eine Zahl ein: "))
        if eingabe < 1 or eingabe > 6:
            print("ABitte geben Sie eine gültige ganze Zahl zwischen 1 und 6 an.")
        else:
            break
    except Exception as e:
        print("Bitte geben Sie eine gültige ganze Zahl zwischen 1 und 6 an.")
        print(e, type(e))
```
