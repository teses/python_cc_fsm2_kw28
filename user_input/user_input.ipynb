# Typecasting der Eingabe ist wichtig um mit den Werten richtig umzugehen.

```python
eingabe = input("Bitte geben Sie eine Zahl ein: ")
print("Die eingegebene Zahl lautet: " + eingabe)

eingabe2 = input("Bitte geben Sie noch eine Zahl ein: ")
print("Die eingegebene Zahl lautet: " + eingabe2)

ergebnis = int(eingabe) + int(eingabe2)
print("Das Ergebnis lautet: " + str(ergebnis))
```
