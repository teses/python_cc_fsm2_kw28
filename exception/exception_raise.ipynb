# Nutzen: selber auf eigene Fehler reagieren

```python
def quadratwurzel_heron(zahl, genauigkeit=0.000001):
    if zahl < 0:
        raise ValueError("Negative Zahlen haben keine reelle Quadratwurzel.")

    if zahl == 0:
        return 0

    schaetzung = zahl

    while abs(schaetzung * schaetzung - zahl) > genauigkeit:
        schaetzung = (schaetzung + zahl / schaetzung) / 2

    return schaetzung


def quadratwurzel(zahl):
    if zahl < 0:
        raise ValueError("Die Zahl muss größer oder gleich 0 sein.")
    return zahl ** 0.5



print(quadratwurzel(25))
print(quadratwurzel(10))

try:
    print(quadratwurzel(-25))
except Exception as e:
    print(e)

```
