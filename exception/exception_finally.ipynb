# Der finally Block

Der finally Block wird auf jeden Fall ausgeführt.

Man könnte es auch als Clean-Up verwenden.

```python
zahl = "Hallo"

try:
    erg = 4 + zahl
except Exception as err:
    print("Es ist ein Fehler aufgetreten: ", err)
finally:
    print("Das hier steht in finally")
```

```python
try:
    erg = 4 + zahl1
except Exception as err:
    print("Es ist ein Fehler aufgetreten: ", err)
    exit()

# das wird nicht mehr ausgeführt
# Aber achtung! in Jupyther steigt er mit exit() nicht aus
print("Das hier steht in finally")
```

# Beispiel bei Dateien auslesen

```python
# Beispiel um finally zu zeigen
# normalerweise werden wir Dateien anders öffnen

#file = "notexistfile.txt"
file = "existfile.txt"

try:
    f = open(file, "r") # lesend öffnen # flock ?

    try:
        f.write("Lorem Ipsum")
    except:
        print("Beim schreiben in der Datei '" + file + "' ist ein Fehler aufgetreten")
    finally:
        f.close()

except:
    print("Beim öffnen der Datei '" + file + "' ist ein fehler aufgetreten. Bitte prüfen sie ob die Datei existiert!")
```
