# Fehler abfangen

`try/except`

`try` - Versuche den Code auszuführen

`except` - Fehlerbehandlung (Ausnahme)

Was kann man bei der Ausnahme machen?
- Meldung anzeigen (Menschenlesbarer, ausführlicher, NICHT so kryptisch wie die original Meldung.
    - Gute Fehlermeldungen?
        - Was für ein Fehler ist aufgetreten?
        - Warum ist der Fehler aufgetreten?
        - Wie kann ich das lösen?
- Fehlermeldungen loggen, protokollieren
- Versenden von Warnmeldungen (z.B. per Email)
- Sachen rückgängig machen (Transaktionssicherheit)

# Fehlerklassen

| Fehlerklasse          | Beschreibung                                |
| --------------------- | ------------------------------------------- |
| `Exception`           | Basisklasse fast aller Fehler               |
| `SyntaxError`         | Fehler im Quellcode                         |
| `IndentationError`    | Falsche Einrückung                          |
| `NameError`           | Variable oder Funktion existiert nicht      |
| `TypeError`           | Falscher Datentyp                           |
| `ValueError`          | Ungültiger Wert bei richtigem Datentyp      |
| `IndexError`          | Listenindex außerhalb des gültigen Bereichs |
| `KeyError`            | Schlüssel im Dictionary nicht vorhanden     |
| `AttributeError`      | Objekt besitzt Attribut oder Methode nicht  |
| `FileNotFoundError`   | Datei wurde nicht gefunden                  |
| `PermissionError`     | Keine Berechtigung für den Zugriff          |
| `ZeroDivisionError`   | Division durch 0                            |
| `ImportError`         | Fehler beim Importieren eines Moduls        |
| `ModuleNotFoundError` | Modul wurde nicht gefunden                  |
| `OverflowError`       | Zahl zu groß für die Berechnung             |
| `MemoryError`         | Nicht genügend Arbeitsspeicher              |
| `RecursionError`      | Maximale Rekursionstiefe überschritten      |
| `RuntimeError`        | Allgemeiner Laufzeitfehler                  |
| `NotImplementedError` | Methode noch nicht implementiert            |
| `KeyboardInterrupt`   | Programm durch den Benutzer abgebrochen     |


# Fehler allgemein abfangen

Nicht empfohlen, weil man dadurch evtl. wichtige Fehler übersieht.
Problem: man muss die Fehlerklasse wissen

```python
try:
    #erg = 10 * (1 / 0) # ZeroDivisionError
    #erg = 4 * zahl # NameError
    #erg = '2' + 2 # TypeError
    x = int("ABC") # ValueError
except Exception as e:
    print("error")
    print("Fehlerklasse: ", type(e).__name__)
    print("Ein Fehler ist aufgetreten:", e)
```

# Bestimmte Fehler und mehrere Fehler unterschiedlich abfangen

```python
try:
    #erg = 10 * (1 / 0) # ZeroDivisionError
    #erg = 4 * zahl # NameError
    #erg = '2' + 2 # TypeError
    x = int("ABC") # ValueError
except ZeroDivisionError as e:
    print("ZeroDivisionError: Durch 0 kann man nicht teilen!", e)
except NameError as e:
    print("NameError: Fehler", e)
except TypeError as e:
    print("TypeError: Fehler", e)
except ValueError as e:
    print("Du musst eine Zahl eingeben!")
except Exception as e:
    print("error")
    print("Fehlerklasse: ", type(e).__name__)
    print("Ein Fehler ist aufgetreten:", e)
```

# Mehrere Fehlerklassen gleichzeitig abfangen

```python
try:
    #erg = 10 * (1 / 0) # ZeroDivisionError
    erg = 4 * zahl # NameError
    #erg = '2' + 2 # TypeError
    #x = int("ABC") # ValueError
except (NameError, TypeError) as e:
    print("Fehler", e, type(e).__name__)
except Exception as e:
    print("error")
    print("Fehlerklasse: ", type(e).__name__)
    print("Ein Fehler ist aufgetreten:", e)
```
