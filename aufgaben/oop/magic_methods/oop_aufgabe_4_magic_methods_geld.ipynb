# Aufgabenstellung: Eine eigene Geld-Klasse in Python

## Ausgangssituation

Du sollst eine kleine Python-Klasse entwickeln, mit der Geldbeträge objektorientiert dargestellt und verarbeitet werden können.

Ein Geldbetrag besteht aus:

* einem **Betrag**, z. B. `25.50`
* einer **Währung**, z. B. `"EUR"`

Ziel ist es, dass sich Objekte der Klasse `Geld` möglichst natürlich verwenden lassen.

Zum Beispiel soll später folgender Code funktionieren:

```python
geld1 = Geld(100, "EUR")
geld2 = Geld(50, "EUR")

print(geld1)
print(geld1 + geld2)
print(geld1 == geld2)
print(geld1 < geld2)
```

## Teil 4 – Geldbeträge addieren

Erweitere die Klasse so, dass zwei Geldobjekte mit `+` addiert werden können.

Beispiel:

```python
geld1 = Geld(100, "EUR")
geld2 = Geld(50, "EUR")

ergebnis = geld1 + geld2

print(ergebnis)
```

Erwartete Ausgabe:

```text
150 EUR
```

Dafür muss eine spezielle Methode für den Operator `+` implementiert werden.

### Wichtige Regel

Geldbeträge unterschiedlicher Währungen dürfen **nicht** addiert werden.

```python
geld1 = Geld(100, "EUR")
geld2 = Geld(50, "USD")

geld1 + geld2
```

In diesem Fall soll ein `ValueError` mit einer verständlichen Fehlermeldung ausgelöst werden.

## Teil 6 – Geldbeträge vergleichen

Implementiere den Vergleichsoperator `==`.

Folgende Vergleiche sollen funktionieren:

```python
geld1 = Geld(100, "EUR")
geld2 = Geld(100, "EUR")
geld3 = Geld(50, "EUR")

print(geld1 == geld2)
print(geld1 == geld3)
```

Erwartete Ausgabe:

```text
True
False
```

Dabei müssen sowohl der **Betrag als auch die Währung** berücksichtigt werden.

Auch folgende Situation soll korrekt behandelt werden:

```python
Geld(100, "EUR") == Geld(100, "USD")
```

```text
False
```

Ein Geldobjekt soll sich folgendermaßen verhalten:
