# Teil 4 – Lagerbestände addieren

Nun soll der `+`-Operator für Produkte verwendet werden.

Gegeben sind zwei Produkte:

```python
p1 = Produkt("Laptop", 1000.00, 5)
p2 = Produkt("Laptop", 1200.00, 3)
```

Folgender Ausdruck soll möglich sein:

```python
p3 = p1 + p2
```

Das Ergebnis soll ein neues `Produkt`-Objekt sein:

```text
Name: Laptop
Menge: 8
```

Der Preis kann dabei beispielsweise vom ersten Produkt übernommen werden.

Die beiden Lagerbestände sollen also addiert werden:

```text
5 + 3 = 8
```

### Wichtige Regel

Die Lagerbestände dürfen **nur bei gleichen Produkten** addiert werden.

Folgender Code:

```python
p1 = Produkt("Laptop", 1000.00, 5)
p2 = Produkt("Monitor", 300.00, 3)

p1 + p2
```

soll einen `ValueError` auslösen.

Verwende eine verständliche Fehlermeldung.

### Hinweis

Für die Implementierung des `+`-Operators benötigst du eine magische Methode.

Recherchiere:

```text
__add__
```

# Teil 5 – Produkte nach dem Preis vergleichen

Produkte sollen anhand ihres Preises verglichen werden können.

Gegeben:

```python
p1 = Produkt("Laptop", 1000.00, 5)
p2 = Produkt("Monitor", 1200.00, 3)
```

Folgender Vergleich soll möglich sein:

```python
print(p1 < p2)
```

Das Ergebnis soll sein:

```text
True
```

Denn:

```text
1000 < 1200
```

Implementiere dafür die entsprechende magische Methode.

### Zusatz

Auch der Gleichheitsoperator `==` soll funktionieren.

```python
print(p1 == p2)
```

Der Vergleich soll sich dabei ausschließlich auf den **Preis** beziehen.

Beispiel:

```python
p1 = Produkt("Laptop", 1000.00, 5)
p2 = Produkt("Monitor", 1000.00, 10)

print(p1 == p2)
```

Ergebnis:

```text
True
```

Obwohl es unterschiedliche Produkte sind, haben beide denselben Preis.

Implementiere dafür:

```text
__eq__(self, other) – ==
__ne__(self, other) – !=
__lt__(self, other) – <
__le__(self, other) – <=
__gt__(self, other) – >
__ge__(self, other) – >=
```

# Teil 6 – Preis mit einem Faktor multiplizieren

Nun soll der Preis eines Produkts mit einem Faktor multipliziert werden können.

Beispiel:

```python
produkt = Produkt("Laptop", 1000.00, 5)

neues_produkt = produkt * 1.19
```

Das Ergebnis soll ein **neues Produktobjekt** sein.

Der Preis soll dabei berechnet werden als:

```text
1000 × 1.19 = 1190
```

Die Lagermenge soll unverändert bleiben.

Damit soll beispielsweise folgender Code funktionieren:

```python
print(produkt * 1.19)
```

Erwartete Ausgabe:

```text
Laptop: 1190.0€, 5 Stück auf Lager
```

Der Faktor soll nur eine Zahl sein (`int` oder `float`).

Implementiere dafür die magische Methode:

```text
__mul__
```
