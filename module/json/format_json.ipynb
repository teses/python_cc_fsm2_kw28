Dictionary aus einen JSON String erstellen

```python
import json

text = '{"name": "Thomas", "alter": 48}'
daten = json.loads(text) # in dictionary umwandeln

print(daten["name"])
```

Aus einem Dictionary JSON erstellen

```python
import json

daten = {
    "name": "Thomas",
    "alter": 48,
    "skills": ["Python", "SQL", "Docker"]
}
json_string = json.dumps(daten, indent=4)  # mit ident 4 wird der string eingerückt

print(json_string)
```
