```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.arange(0, 10, 1)

# Y-Werte
m = 2
d = 0

y = m * x + d

print("x", x, type(x), id(x))
print("y",y , type(y), id(y))

# plot
plt.plot(x,y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.arange(0, 10, 0.1)

# Y-Werte
y = x ** 2

print("x", x, type(x), id(x))
print("y",y , type(y), id(y))

# plot
plt.plot(x,y)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# x-Werte
x = np.arange(0, 2*np.pi, 0.1)

# Y-Werte
y = np.sin(x)

print("x", x, type(x), id(x))
print("y",y , type(y), id(y))

# plot
plt.plot(x,y)
plt.show()
```
