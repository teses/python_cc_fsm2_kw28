```python
import pandas as pd

df = pd.read_csv("../../data/missing_data.csv")
```

# Analyse

```python
# Fehlende Werte finden
print(df.isnull())
print(df.isnull().sum())
```

# Manipulation

```python
# Zeilen mit fehlenden Werten entfernen
df_clean = df.dropna()
print(df_clean)

# Menge der richtigen Zeilen
print(len(df.dropna()))
```

# fehlende Werte ersetzen

```python
df2 = df.fillna(100)  # mit 100 ersetzen bei allen Spalten
print(df2)
```

# Fehlende Werte einer Spalte ersetzen

```python
mean = df["alter"].mean()
median = df["alter"].median()
print(mean, median)

# copy erstellen
df2 = df.copy() # copy nicht vergessen

# Fehlende Werte im alter ersetzen mit durchschnitt
df2["alter"] = df["alter"].fillna(median)
# stadt cleanen
df2["stadt"] = df["stadt"].fillna("")


print(df2)
```
