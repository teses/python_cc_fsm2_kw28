```python
from sqlalchemy import create_engine
from sqlalchemy.types import Integer, String, Float, DateTime
import pandas as pd

#
connection = create_engine(
    "mysql+pymysql://root:@localhost:3306/comcave_etl",
    echo=True  # debug ausgabe
)


file = "../../data/missing_data.csv"
df = pd.read_csv(file)

count = df.to_sql(
    "missing",
    con=connection,
    if_exists="replace",  # replace , append
    index=True,
    dtype={
        "name": String(50)
    }
)
print(f"importiert: {count}")
```

```python
from sqlalchemy import create_engine
from sqlalchemy.types import Integer, String, Float, DateTime
import pandas as pd

#
connection = create_engine(
    "mysql+pymysql://root:@localhost:3306/comcave_etl",
    echo=True  # debug ausgabe
)

# df
df = pd.DataFrame({
    "Name": ["Max", None, "Tom"],
    "Alter": [25, 22, None],
    "Geburtstag": [
        pd.NaT,
        pd.Timestamp("2001-05-10"),
        pd.Timestamp("2004-08-15")
    ],
    "Mein Test": ["Hallo", None, "NaN"],
})

count = df.to_sql(
    "missing2",
    con=connection,
    if_exists="replace",  # replace , append
    index=True,
    dtype={
        "name": String(50)
    }
)
print(f"importiert: {count}")
```
