```python
from sqlalchemy import create_engine
import pandas as pd

#
connection = create_engine(
    "mysql+pymysql://root:@localhost:3306/comcave_etl",
    echo=True  # debug ausgabe
)


# Daten als JSON
daten = [
    {"Name": "Max", "Alter": 25, "Stadt": "Berlin"},
    {"Name": "Anna", "Alter": 30, "Stadt": "Hamburg"},
    {"Name": "Tom", "Alter": 22, "Stadt": "Köln"}
]
df = pd.DataFrame(daten)

count = df.to_sql(
    "personen",
    con=connection,
    if_exists="replace",  # replace , append
    index=False
)
print(f"importiert: {count}")
```
