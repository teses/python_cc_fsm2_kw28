```python
import pandas as pd

file = "../../data/sales_noheader.csv"
df = pd.read_csv(
    file,
    sep=",",
    header=None, # Datei hat keine Spaltennamen
    names=["Datum","Produkt","Kategorie","Preis","Menge","Region"],
)

#print(df)
print(df.to_string())
```

```python
import pandas as pd

file = "../../data/sales_noheader.csv"
df = pd.read_csv(
    file,
    sep=",",
    header=None, # Datei hat keine Spaltennamen
    names=["Datum","Produkt","Kategorie","Preis","Menge","Region"],
    usecols=["Datum","Produkt"]
)

#print(df)
print(df.to_string())
```
