# Eine JSON Datei lesen

Die JSON Datei direkt mit Pandas zu lesen eignet sich nur wenn diese nicht zu stark verschachtelt ist
Bei verschachtelten JSON ist es meist besser diese zu "flatten"


```python
import pandas as pd
file = "../../data/krankenkassen.json"
df = pd.read_json(file)
print(df)
```

```python
print(df["features"].iloc[0]["properties"]["bezeichnung"])

for d in df["features"].iloc:
    print(d["properties"]["bezeichnung"])
```

```python
import pandas as pd
import json

file = "../../data/krankenkassen.json"
with open(file, "r") as f:
    data = json.load(f)


df = pd.json_normalize(
    data["features"],
    sep ="_"
)

#print(df.head())

for d in df.iloc:
    print(d)

```
