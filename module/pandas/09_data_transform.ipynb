```python
daten = {
    "Name": ["Anna", "Tom", "Lisa"],
    "Alter": [22, 30, 27],
    "Gewicht": [50.2, 78.2, 66.5],
    "Status": [0, 0, 1]
}
df = pd.DataFrame(daten)

# Neue Spalte erstellen
df["AlterIn5Jahren"] = df["Alter"] + 5

# Werte berechnen
df["GewichtMal10"] = df["Gewicht"] * 10

# Werte mit Funktion ändern
df["Alter1"] = df["Alter"].apply(lambda x : x + 1)

def testFunc(x):
    return x + 2
df["Alter2"] = df["Alter"].apply(testFunc)

# mapping
df["Status2"] = df["Status"].map({
    0: "Inaktiv",
    1: "Aktiv"
})
print(df)
```
