# Daten bereinigen

```python
import pandas as pd

df = pd.read_csv("../../data/kunden_dirty_data.csv")

# spalten bereinigen
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()

# Daten bereinigen
df['name'] = df['name'].str.strip() # leerzeichen vorne und hinten weg
df["email"] = df["email"].str.strip()
df["land"] = df["land"].str.strip()
df["umsatz"] = df["umsatz"].replace("-", pd.NA)

print(df)
```

```python
import pandas as pd

df = pd.read_csv("../../data/kunden_dirty_data.csv")

for column in df.columns.to_list():
    if str(df[column].dtype) == "object":
        df[column] = df[column].str.strip()


print(df)
```

# Datumswerte konvertieren - Vorsicht kann die Daten auch zerstören

```python
import pandas as pd

df = pd.read_csv("../../data/kunden_dirty_data.csv")

# errors="coerce" macht ungültige Daten zu NaT.
df["geburtsdatum"] = pd.to_datetime(
    df["geburtsdatum"],
    errors="coerce",
    format="mixed",
    dayfirst=True
)

df["registriert_am"] = pd.to_datetime(
    df["registriert_am"],
    errors="coerce",
    format="mixed",
    dayfirst=True
)

print(df)
```

# Zahlenformate bereinigen

```python
import pandas as pd

df = pd.read_csv("../../data/kunden_dirty_data.csv")

df["telefon"] = (
    df["telefon"]
    .astype(str)
    .str.replace(" ", "")
    .str.replace("-", "")
    .str.replace("+", "00")
)
print(df)
```

# Datentypen ändern
