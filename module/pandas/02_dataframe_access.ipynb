```python
import pandas as pd
# Liste mit Dictionaries
personen = [
    {"Name": "Max", "Alter": 25, "Stadt": "Berlin"},
    {"Name": "Anna", "Alter": 30, "Stadt": "Hamburg"},
    {"Name": "Tom", "Alter": 22, "Stadt": "Köln"}
]
df = pd.DataFrame(personen)
#print(df)
#print(df.to_string()) # zeigt alles an
```

# Eine Spalte anzeigen

```python
print(df["Stadt"])
```

# mehrere Spalten

```python
print(df[["Name", "Stadt"]])
```

# Zugriff auf die Zeile

```python
print(df.iloc[0])
print(df.iloc[0]["Name"])
```

# Zugriff auf mehrere Zeilen mit Slicing

```python
print(df.iloc[0:2])
print(df.iloc[0:2:2])
print(df.iloc[::-1])
```

# Neue Spalte hinzufügen

```python
df["Gehalt"] = [3000, 3500, 4000]
print(df)
```
