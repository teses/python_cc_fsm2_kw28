# CSV Daten auslesen

```python
import pandas as pd

file = "../../data/sales_varianz.csv"
df = pd.read_csv(file)

#print(df)
print(df.to_string())
```

# Seperator und encoding

```python
import pandas as pd

file = "../../data/sales_semikolon.csv"
df = pd.read_csv(
    file,
    sep=";",
    encoding="utf-8"
)

#print(df)
print(df.to_string())
```

```python
import pandas as pd

file = "../../data/sales_noheader.csv"
df = pd.read_csv(
    file,
    sep=",",
    header=None, # Datei hat keine Spaltennamen
    names=["Datum","Produkt","Kategorie","Preis","Menge","Region"],
)

#print(df)
print(df.to_string())
```

```python
import pandas as pd

file = "../../data/sales_noheader.csv"
df = pd.read_csv(
    file,
    sep=",",
    header=None, # Datei hat keine Spaltennamen
    names=["Datum","Produkt","Kategorie","Preis","Menge","Region"],
    usecols=["Datum","Produkt"]
)

#print(df)
print(df.to_string())
```

# Fehlerhafte Werte als NaN

```python
import pandas as pd

file = "../../data/missing_data.csv"
df = pd.read_csv(
    file,
    na_values="-"
)

#print(df)
print(df.to_string())
```

# Datentypen angeben

```python
import pandas as pd

file = "../../data/sales_varianz.csv"
df = pd.read_csv(
    file,
    dtype={
        "Preis" : str,
        "Menge" : str
    }

)

print(df)
print(df.info())
```
