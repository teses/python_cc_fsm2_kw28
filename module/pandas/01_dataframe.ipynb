# DataFrame

- Im Prinzip eine Tabelle
- Unterschied zu normalen Tabellen: Die Tabelle ist in Spalten aufgeteilt

# DataFrame erstellen aus einer Zeilenorientierter Struktur

```python
import pandas as pd
# Liste mit Dictionaries
personen = [
    {"Name": "Max", "Alter": 25, "Stadt": "Berlin"},
    {"Name": "Anna", "Alter": 30, "Stadt": "Hamburg"},
    {"Name": "Tom", "Alter": 22, "Stadt": "Köln"}
]
df = pd.DataFrame(personen)
print(df)
print(df.to_string()) # zeigt alles an
```

# DataFrame erstellen aus einer Spaltenorientierter Struktur

```python
# Dictionary aufgebaut nach Spalten
daten = {
    "Name": ["Anna", "Tom", "Lisa"],
    "Alter": [22, 30, 27]
}
df = pd.DataFrame(daten)
print(df)
```

# DataFrame erstellen mit einer verschachtelten Liste (Zeilen + Spalten)

```python
daten = [
    ["Max", 25, "Berlin"],
    ["Anna", 30, "Hamburg"],
    ["Tom", 22, "Köln"]
]
df = pd.DataFrame(daten, columns=["Name", "Alter", "Stadt"])
print(df)
```
