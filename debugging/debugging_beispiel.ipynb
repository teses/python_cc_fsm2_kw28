```python
def finde_maximum(zahlen_liste):
    max_wert = 0  # <--- Hier Start-Breakpoint setzen
    for zahl in zahlen_liste:
        if zahl > max_wert:
            max_wert = zahl
    return max_wert

meine_zahlen = [-5, -2, -9, -1, -3]
ergebnis = finde_maximum(meine_zahlen)
print(f"Das Maximum ist: {ergebnis}")
```

- Nutze Step Over, um durch die Schleife zu gehen.
