Klasse ohne @property

Klassische Gtter/Setter

```python
class Person:

    def __init__(self, name=""):
        self.__name = name


    def get_name(self):
        return self.__name


    def set_name(self, name):
        self.__name = name


p1 = Person("Max")
print(p1.get_name())
p1.set_name("Franz")
print(p1.get_name())
```

Das @property-Decorator in Python wird verwendet, um Methoden einer Klasse wie Attribute zu behandeln.
Damit kann man getter, setter und deleter für ein Attribut definieren,
ohne dass man eine explizite get_...() oder set_...() Methode schreiben muss.

```python
class Person:

    def __init__(self, name=""):
        #self.__name = name
        self.name = name

    @property
    def name(self):  # Getter
        print(f"ich bin die getter methode und gebe {self.__name} zurück")
        return self.__name

    @name.setter
    def name(self, value): # Setter
        print(f"ich bin die setter methode und setzte {value}")
        self.__name = value


p = Person("Max")
print(p.name)
p.name = "Helmut"
print(p.name)
```
