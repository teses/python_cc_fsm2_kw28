```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person: {self.name} ({self.age})"

    def meine_methode(self, a):
        return a

class Student(Person):

    def __init__(self, name, age, year):
        #Person.__init__(self, name, age) # funktioniert
        super().__init__(name,age)    #  moderner / bevorzugt
        self.year = year

    # Diese Methode überschreibt die __str__ Methode aus der Klasse Person
    def __str__(self):
        return f"Person: {self.name} ({self.age}) ({self.year})"

    def meine_methode_in_student(self, a):
        return a

```
