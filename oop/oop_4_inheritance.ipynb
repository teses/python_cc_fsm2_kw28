Vererbung (Inheritance) ist ein Grundkonzept der objektorientierten Programmierung.

 Eine Klasse kann Eigenschaften und Methoden einer anderen Klasse übernehmen und erweitern.

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person: {self.name} ({self.age})"


class Student(Person):

    def __init__(self, name, age, year):
        #Person.__init__(self, name, age) # funktioniert
        super().__init__(name,age)    #  moderner / bevorzugt
        self.year = year

    # Diese Methode überschreibt die __str__ Methode aus der Klasse Person
    def __str__(self):
        return f"Person: {self.name} ({self.age}) ({self.year})"



meinePerson = Person("Max", 50)
print(meinePerson)

meinStudent = Student("Hans", 44, 2005)
print(meinStudent)

```
