# Factory Methode in der Abstrakten Oberklasse

```python
from abc import ABC, abstractmethod

# KLasse Fahrzeug ist Abstrakt
class Fahrzeug(ABC):

    @staticmethod
    def factory(typ):
        if typ == "auto":
            return Auto()
        elif typ == "motorrad":
            return Motorrad()
        elif typ == "fahrrad":
            return Fahrrad()
        else:
            raise ValueError(f"Fahrzeugtyp unbekannt: {typ}")

    @abstractmethod
    def fahren(self):
        pass

class Auto(Fahrzeug):
    def fahren(self):
        print("Auto fährt")

class Motorrad(Fahrzeug):
    def fahren(self):
        print("Motorrad fährt")

class Fahrrad(Fahrzeug):
    def fahren(self):
        print("Fahrrad fährt")

```

# Implementierung alt (ohne Factory)

```python
entry = "auto"

if entry == "auto":
    myObj = Auto()
elif entry == "motorrad":
    myObj = Motorrad()
elif entry == "fahrrad":
    myObj = Fahrrad()

myObj.fahren()
```

# Implementierung mit Factory

```python
entry = "auto"

myObj = Fahrzeug.factory(entry) # Factory in der Abstrakten Oberklasse
#myObj = FahrzeugFabrik(entry)  # Factory Methode

myObj.fahren()
```
