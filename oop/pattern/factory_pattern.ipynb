Wir bauen eine Funktion/Klasse die entscheidet anhand einer Eingabe, welche konkrete Klasse instanziert wird.

**Klassen**
- Fahrzeug
  - Auto
  - Motorrad
  - Fahrrad

Die Fabrik wird entweder in einer eigenen Klasse ausgesiedelt, oder in der Abstrakten Oberklasse
- FahrzeugFabrik

```python
# so ist es implementiert an hunderten Stellen im Code

entry = "auto"

if entry == "auto":
    myObj = Auto()
elif entry == "motorrad":
    myObj = Motorrad()
elif entry == "fahrrad":
    myObj = Fahrrad()

myObj.fahren()
```

# Simple Factory Funktion

```python
def FahrzeugFabrik(typ):
    if typ == "auto":
        myObj = Auto()
    elif typ == "motorrad":
        myObj = Motorrad()
    elif typ == "fahrrad":
        myObj = Fahrrad()

    return myObj


# Implementierung
entry = "auto"
myObj = FahrzeugFabrik(entry)
myObj.fahren()
```
