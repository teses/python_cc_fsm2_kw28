# Dataclass

Eine Dataclass ist eine spezielle Python-Klasse, mit der man Daten einfach und übersichtlich speichern kann.

Damit lassen sich Klassen erstellen, die hauptsächlich dazu dienen, Daten strukturiert zu speichern.

Verwendet wird dafür das Dekorator-Pattern @dataclass aus dem Modul dataclasses.

Ohne Dataclass müsste man z. B. den Konstruktor selbst schreiben:

```python
class Todo:
    def __init__(self, id, title, completed):
        self.id = id
        self.title = title
        self.completed = completed


todo = Todo(1, "Python lernen", False)
print(todo.title)
```

Mit dataclass geht es deutlich kürzer:

```python
from dataclasses import dataclass

@dataclass
class Todo:
    id: int
    title: str
    completed: bool


todo = Todo(id=1, title="Python lernen", completed=True)
print(todo.id , todo.title)
```
