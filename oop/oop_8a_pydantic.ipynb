# Pydantic – kurz erklärt

Pydantic ist eine Python-Bibliothek, mit der man Datenstrukturen definieren und Daten automatisch überprüfen (validieren) kann.

Achtung!
Pydantic und dataclasses sehen auf den ersten Blick ähnlich aus, weil beide Python-Klassen mit typisierten Attributen ermöglichen. Der entscheidende Unterschied ist aber:

`dataclasses` sind hauptsächlich für die Strukturierung von Python-Daten gedacht. `Pydantic` ist zusätzlich für Validierung und Datenkonvertierung gedacht.


```python
from pydantic import BaseModel

class Todo(BaseModel):
    id: int
    title: str
    completed: bool
```

Damit wird festgelegt:

- id muss eine Zahl sein
- title muss ein Text sein
- completed muss ein Boolean (True/False) sein

Beim Erzeugen eines Objekts überprüft Pydantic die Daten:

```python
todo = Todo(id=5, title="Python lernen", completed=True)
print(todo.id , todo.title)
```

Beim Erzeugen eines Objekts überprüft Pydantic die Daten:

```python
todo = Todo(id="hallo", title="Python lernen", completed=True)
print(todo.id , todo.title)
```
