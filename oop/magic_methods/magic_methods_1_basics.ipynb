In Python sind magische Methoden (oft auch „Dunder-Methoden“ genannt,
weil sie mit doppelten Unterstrichen __ beginnen und enden) spezielle Methoden,
die bestimmte Standardoperationen eines Objekts definieren oder überschreiben.
Sie ermöglichen es, dass eigene Klassen sich wie eingebaute Typen verhalten.

# Konstruktor und Initialisierung

- __new__(cls, ...) – Erstellt eine neue Instanz (selten überschrieben).
- __init__(self, ...) – Initialisiert eine Instanz nach der Erstellung.

```python
class Person:
    def __init__(self, name):
        self.name = name

p = Person("Alice")
print(p.__dict__)
```

# Darstellung / String-Konvertierung
- __str__(self) – Für menschenlesbare Darstellung (print(obj)).
- __repr__(self) – Für Entwickler - zeigt wie Objekt erstellt wurde (repr(obj)).

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Person: {self.name}"

    def __repr__(self):
        return f"Person({self.name!r})"


p = Person("Alice")
print(p)
print(repr(p))
```
