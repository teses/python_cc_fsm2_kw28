
**Container-Verhalten**
```
__len__(self) – für len(obj)
__getitem__(self, key) – Zugriff per obj[key]
__setitem__(self, key, value) – Zuweisung obj[key] = value
__delitem__(self, key) – Löschen del obj[key]
__iter__(self) – Iterator starten (z. B. in for)
__next__(self) – Nächstes Element liefern
__contains__(self, item) – in
```

````python
class Bibliothek:

    def __init__(self):
        self.buecher = []

    def append(self, value):
        self.buecher.append(value)

    # len(bibliothek)
    def __len__(self):
        return len(self.buecher)

    # bibliothek[key]
    def __getitem__(self, key):
        return self.buecher[key]

    # bibliothek[key] = value
    def __setitem__(self, key, value):
        self.buecher[key] = value

    # del bibliothek[key]
    def __delitem__(self, key):
        del self.buecher[key]

    # for buch in bibliothek
    def __iter__(self):
        return iter(self.buecher)

    # item in bibliothek
    def __contains__(self, item):
        return item in self.buecher

````

````python
# Bibliothek erzeugen
bibliothek = Bibliothek()

# Bücher hinzufügen
bibliothek.append("Python")
bibliothek.append("SQL")
bibliothek.append("FastAPI")

# __len__()
print(len(bibliothek))

# __getitem__()
print(bibliothek[0])
print(bibliothek[1])

# __setitem__()
bibliothek[1] = "Datenbanken"
print(bibliothek[1])

# __delitem__()
del bibliothek[2]
print(len(bibliothek))

# __iter__()
for buch in bibliothek:
    print(buch)

# __contains__()
print("Python" in bibliothek)
print("FastAPI" in bibliothek)
````
