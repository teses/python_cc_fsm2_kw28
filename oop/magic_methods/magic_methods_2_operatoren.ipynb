# Operatorüberladung

**Arithmetische Operatoren**
```
__add__(self, other) – Überladung von +.
__sub__(self, other) – Überladung von -.
__mul__(self, other) – Überladung von *.
__truediv__(self, other) – Überladung von /.
__floordiv__(self, other) – //
__mod__(self, other) – %
__pow__(self, other) – **
```

**Vergleichsoperatoren**
```
__eq__(self, other) – ==
__ne__(self, other) – !=
__lt__(self, other) – <
__le__(self, other) – <=
__gt__(self, other) – >
__ge__(self, other) – >=
```


````python
class Vektor:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vektor({self.x}, {self.y})"


    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Vektor(x, y)


    def __sub__(self, other):
        x = self.x - other.x
        y = self.y - other.y
        return Vektor(x, y)


    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return Vektor(self.x * other, self.y * other)
        return Vektor(self.x * other.x, self.y * other.y)


    def __truediv__(self, other):
        if isinstance(other, (int, float)):
            return Vektor(self.x / other, self.y / other)
        return Vektor(self.x / other.x, self.y / other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

````

````python
v1 = Vektor(3, 4)
print(repr(v1))

v2 = Vektor(1, -1)
print(repr(v2))

v_a =  v1 + v2
print("__add__", v_a)  # __add__
print(repr(v_a))

v_s =  v1 - v2
print("__sub__", v_s)  #
print(repr(v_s))
````

````python
v1 = Vektor(6, 4)
print(repr(v1))

v2 = Vektor(2, 2)
print(repr(v2))

print("__sub__", v1 - v2)  #
print("__mul__", v1 * 3)  #
print("__mul__", v1 * v2)  #
print("__truediv__", v1 / 2)  #
print("__truediv__", v1 / v2)  #

print("__eq__", v1 == Vektor(6, 4))  #
````
