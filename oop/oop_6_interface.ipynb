# Interface

wie baue ich ein Interface in Python

```python
from abc import ABC, abstractmethod

# Das ist das Interface
class LoggerInterface(ABC):

    @abstractmethod
    def log(self):
        pass


class User():

    def __init__(self, name):
        self.name = name


class Person(User, LoggerInterface):

    def log(self):
        print("logging logging....")




p = Person("Max")
p.log()
```
