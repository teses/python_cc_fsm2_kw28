```python
class Person :

    # Dies sind Klassenattribute in allen objekten verfügbar
    personCounts = 0

    # Konstruktor
    def __init__(self, firstname, lastname):
        # Normale Attribute die zur Laufzeit definiert werden
        self.firstname = firstname
        self.lastname = lastname

    def __str__(self):
        out = f"Person Klasse: \r\n firstname: {self.firstname} \r\n lastname: {self.lastname} \r\n"
        out += f" personCounts: {self.personCounts}"
        return out


p1 = Person("Max", "Mustermann")
p1.personCounts = 1
p1.personCounts += 1
print(p1)
print(Person.personCounts)

##################################
Person.personCounts = 3

##################################
p2 = Person("Susi", "Sauer")
p2.personCounts += 1
print(p2)

# Klassenattribute
# Achtung Klassenatribute und Objektattribute sind nicht die selben Speicherplätze
# daher sollten die namen nie gleich sein
print(Person.personCounts)

```
