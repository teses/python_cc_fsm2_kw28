Wenn man von mehreren Klassen erbt, sollte man mit Methoden aufpassen die die Methoden aus der Oberklasse überschreiben

Hier ein Beispiel wo eine Klasse von zwei anderen erbt und die Methode überschreibt

```python
class Person:
    def info(self):
        print("Person")


class Student:
    def info(self):
        print("Student")


class Azubi(Student, Person):
    def info(self):
        print("Azubi")
        super().info()



p1 = Person()
p1.info()

print("-"*20)
azubi = Azubi()
azubi.info()
```
