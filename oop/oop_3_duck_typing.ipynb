# Duck typing

„Wenn es aussieht wie eine Ente, läuft wie eine Ente und quakt wie eine Ente, dann ist es wahrscheinlich eine Ente.“

Wenn ein Objekt die benötigten Methoden besitzt, kann es verwendet werden – unabhängig davon, zu welcher Klasse es gehört.

```python
class Tier:
    pass


class Hund(Tier):
    def lautGeben(self):
        print("Wuff!")


class Katze(Tier):
    def lautGeben(self):
        print("Miau!")


class Auto:
    def lautGeben(self):
        print("Tutut Hup")



daten = [
    Hund(),
    Katze(),
    Hund(),
    Hund(),
    Auto()
]

for obj in daten:
    obj.lautGeben()
```
