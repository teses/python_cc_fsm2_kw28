Man kann in Klassen die erben die Methoden der Oberklasse ganz einfach überschreiben indem man eine MEthode mit gleichen Namen definiert

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person: {self.name} ({self.age})"

    def meine_methode(self):
        return "Methode aus Person"


class Student(Person):

    def __init__(self, name, age, year):
        super().__init__(name, age)
        self.year = year

    # Überschreibt die __str__-Methode aus Person
    def __str__(self):
        return f"Student: {self.name} ({self.age}) ({self.year})"

    # Überschreibt meine_methode aus Person
    def meine_methode(self):
        return "Methode aus Student"


# Objekte erzeugen
person = Person("Holger", 40)
student = Student("Max", 22, 2026)

# Ausgaben
print(person)
print(student)

print(person.meine_methode())
print(student.meine_methode())
```
